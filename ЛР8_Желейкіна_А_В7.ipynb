```python
import numpy as np

# Таблиця значень функції
x_values = np.array([2.4, 2.6, 2.8, 3, 3.2, 3.4, 3.6, 3.8, 4, 4.2, 4.4, 4.6])
y_values = np.array([3.526, 3.782, 3.945, 4.043, 4.104, 4.155, 4.222, 4.331, 4.507, 4.775, 5.159, 5.683])

# Крок таблиці
h = x_values[1] - x_values[0] # h = 0.2

# Обчислення кінцевих різниць
def finite_differences(y):
    n = len(y)
    delta_y = np.zeros((n, n))
    delta_y[:, 0] = y

    for j in range(1, n):
        for i in range(n - j):
            delta_y[i, j] = delta_y[i + 1, j - 1] - delta_y[i, j - 1]

    return delta_y

# Отримуємо таблицю кінцевих різниць
delta_table = finite_differences(y_values)

# Виводимо таблицю кінцевих різниць для наочності
print("Таблиця кінцевих різниць:")
np.set_printoptions(precision=4, suppress=True)
print(delta_table)

# Значення кінцевих різниць для першої та другої похідних
#Змінити індекс відповідно рядка, який використовуєте (нумерація починається з нуля)
delta_y1 = delta_table[3, 1]   # Δy
delta2_y1 = delta_table[3, 2]   # Δ²y
delta3_y1 = delta_table[3, 3]   # Δ³y
delta4_y1 = delta_table[3, 4]   # Δ⁴y
delta5_y1 = delta_table[3, 5]   # Δ⁵y
delta6_y1 = delta_table[3, 6]   # Δ⁶y
delta7_y1 = delta_table[3, 7]   # Δ⁷y
delta8_y1 = delta_table[3, 8]   # Δ⁸y

# Функція для обчислення першої похідної за формулою Ньютона
def first_derivative(delta_y1, delta2_y1, delta3_y1, delta4_y1, delta5_y1, delta6_y1, delta7_y1, delta8_y1, h):
    return (delta_y1 - delta2_y1 / 2 + delta3_y1 / 3 - delta4_y1 / 4 + delta5_y1 / 5 - delta6_y1 / 6 + delta7_y1 / 7 - delta8_y1 / 8) / h

# Функція для обчислення другої похідної за формулою Ньютона
def second_derivative(delta2_y1, delta3_y1, delta4_y1, delta5_y1, delta6_y1, delta7_y1, delta8_y1, h):
    return (delta2_y1 - delta3_y1  + 11*delta4_y1 / 12 - 5*delta5_y1 / 6 + 137*delta6_y1 / 180 - 7*delta7_y1 / 10 + 363*delta8_y1 / 560) / h**2

# Обчислюємо першу та другу похідні в точці x = 3
y1_prime = first_derivative(delta_y1, delta2_y1, delta3_y1, delta4_y1, delta5_y1, delta6_y1, delta7_y1, delta8_y1, h)
y1_double_prime = second_derivative(delta2_y1, delta3_y1, delta4_y1, delta5_y1, delta6_y1, delta7_y1, delta8_y1, h)

# Округляємо до точності 0.001
y1_prime = round(y1_prime, 3)
y1_double_prime = round(y1_double_prime, 3)

# Виводимо результати
print(f"\nПерша похідна y'(0.1) ≈ {y1_prime}")
print(f"Друга похідна y''(0.1) ≈ {y1_double_prime}")
```

```python
import numpy as np

# Таблиця значень функції
x_values = np.array([2.4, 2.6, 2.8, 3.0, 3.2, 3.4, 3.6, 3.8, 4.0, 4.2, 4.4, 4.6])
y_values = np.array([3.526, 3.782, 3.945, 4.043, 4.104, 4.155, 4.222, 4.331, 4.507, 4.775, 5.159, 5.683])

# Крок таблиці
h = x_values[1] - x_values[0]

# Перша похідна за формулою прямої різниці
first_derivative = (y_values[1:] - y_values[:-1]) / h

# Друга похідна (теж за формулою прямої різниці)
second_derivative = (first_derivative[1:] - first_derivative[:-1]) / h

# Обчислення для x = 3
x_target = 3
i = np.where(x_values == x_target)[0][0]

# Для y'(x_i) використовуємо y[i+1] і y[i]
first_derivative_at_x = (y_values[i+1] - y_values[i]) / h

# Для y''(x_i) використовуємо y'(i+1) і y'(i)
second_derivative_at_x = (first_derivative[i+1] - first_derivative[i]) / h

# Виведення результатів
print(f"Для x = {x_target}:")
print(f"Перша похідна y'({x_target}) ≈ {first_derivative_at_x:.5f}")
print(f"Друга похідна y''({x_target}) ≈ {second_derivative_at_x:.5f}")
```

```python
import matplotlib.pyplot as plt

#Таблиця значень функції
x_values = np.array([2.4, 2.6, 2.8, 3, 3.2, 3.4, 3.6, 3.8, 4, 4.2, 4.4, 4.6])
y_values = np.array([3.526, 3.782, 3.945, 4.043, 4.104, 4.155, 4.222, 4.331, 4.507, 4.775, 5.159, 5.683])

#Крок таблиці
h = x_values[1] - x_values[0]

#Обчислення кінцевих різниць
first_derivative = (y_values[1:] - y_values[:-1]) / h          # Перша похідна
second_derivative = (first_derivative[1:] - first_derivative[:-1]) / h  # Друга похідна

#Інтерполяція для гладких ліній
x_fine_y = np.linspace(x_values[0], x_values[-1], 500)
y_fine = np.interp(x_fine_y, x_values, y_values)

# Для похідних треба зменшити діапазон, бо вони коротші
x_data_d1 = x_values[:-1]
x_data_d2 = x_values[:-2]

x_fine_d1 = np.linspace(x_data_d1[0], x_data_d1[-1], 500)
x_fine_d2 = np.linspace(x_data_d2[0], x_data_d2[-1], 500)

first_deriv_fine = np.interp(x_fine_d1, x_data_d1, first_derivative)
second_deriv_fine = np.interp(x_fine_d2, x_data_d2, second_derivative)

# Створення графіків
plt.figure(figsize=(12, 8))

# Функція
plt.subplot(3, 1, 1)
plt.plot(x_fine_y, y_fine, color='red', label='Функція y(x)')
plt.scatter(x_values, y_values, color='black', label='Дані')
plt.title('Функція y(x)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)

# Перша похідна
plt.subplot(3, 1, 2)
plt.plot(x_fine_d1, first_deriv_fine, color='orange', label="Перша похідна y'(x)")
plt.title("Перша похідна y'(x)")
plt.xlabel('x')
plt.ylabel("y'")
plt.legend()
plt.grid(True)

# Друга похідна
plt.subplot(3, 1, 3)
plt.plot(x_fine_d2, second_deriv_fine, color='blue', label="Друга похідна y''(x)")
plt.title("Друга похідна y''(x)")
plt.xlabel('x')
plt.ylabel("y''")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
